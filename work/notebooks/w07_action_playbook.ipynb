```python
from pathlib import Path
import json

repo = Path.cwd()

reason_codes = [
    "visible_model_opportunity",
    "model_decline_risk",
    "declining_with_demand",
    "low_ctr_visible_page",
    "ctr_review_candidate",
    "low_engagement_visible_page",
    "engagement_review_candidate",
    "page_one_decay_risk",
    "stale_visible_page",
]

search_extensions = {
    ".py",
    ".ipynb",
    ".md",
    ".json",
}

print("=" * 70)
print("REASON CODE SOURCE SEARCH")
print("=" * 70)

for reason in reason_codes:
    print("\n" + "=" * 70)
    print("REASON:", reason)
    print("=" * 70)

    found = False

    for path in repo.rglob("*"):
        if not path.is_file():
            continue

        if path.suffix not in search_extensions:
            continue

        if ".git" in path.parts:
            continue

        try:
            if path.suffix == ".ipynb":
                notebook = json.loads(
                    path.read_text(encoding="utf8")
                )

                for cell_number, cell in enumerate(
                    notebook.get("cells", [])
                ):
                    source = "".join(
                        cell.get("source", [])
                    )

                    if reason in source:
                        print(
                            "\nFILE:",
                            path.relative_to(repo)
                        )
                        print(
                            "CELL:",
                            cell_number
                        )

                        lines = source.splitlines()

                        for line_number, line in enumerate(lines):
                            if reason in line:
                                start = max(0, line_number + int(chr(45) + "4"))
                                end = min(
                                    len(lines),
                                    line_number + 5
                                )

                                for context_line in lines[start:end]:
                                    print(context_line)

                                print()

                        found = True

            else:
                text = path.read_text(
                    encoding="utf8",
                    errors="ignore"
                )

                if reason in text:
                    print(
                        "\nFILE:",
                        path.relative_to(repo)
                    )

                    lines = text.splitlines()

                    for line_number, line in enumerate(lines):
                        if reason in line:
                            start = max(0, line_number + int(chr(45) + "4"))
                            end = min(
                                len(lines),
                                line_number + 5
                            )

                            for context_line in lines[start:end]:
                                print(context_line)

                            print()

                    found = True

        except Exception:
            pass

    if not found:
        print("No source definition found")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## Intended use and limits

### Intended use

This system is designed to rank existing content items for human refresh review. It helps a reviewer decide which pages may deserve attention first based on observed performance signals and the validated model score.

The model was evaluated using a client holdout validation strategy. The selected Random Forest achieved precision at 50 of 0.740. The ranked queue should therefore be interpreted as decision support for prioritization, not as an automatic publishing or editing system.

The recommended workflow is:

1. Review higher ranked items first.
2. Read the reason codes attached to each recommendation.
3. Check the page and its editorial context manually.
4. Decide whether a refresh, CTR review, engagement review, expansion, or no action is appropriate.

### Limits

1. The analysis uses the anonymized FlyRank starter dataset containing 30,000 content items. Results describe patterns observed in this dataset and may not generalize to every client or future period.

2. The model identifies associations and ranking signals. It does not establish that refreshing a page will cause traffic, ranking, CTR, or engagement to improve.

3. The target is `is_declining_label`. Leakage related fields such as `trend_direction` and `trend_pct` were not used as model features.

4. Client and content identifiers are pseudonymous and are used for grouping, validation, and identification only. They are not predictive features.

5. The model does not inspect page titles, URLs, domains, client names, or keywords, so important editorial and business context may be missing.

6. Model probability is a ranking signal, not a guarantee that an individual page is declining or should be changed.

7. Recommendations require human review before any content change is made.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human review and the no go list

### Human review

Every recommendation in the ranked queue requires human review before any content change is made.

The reviewer should check:

1. Whether the page is actually declining in a meaningful way.

2. Whether the reason codes match the current page context.

3. Whether traffic and search visibility are high enough to justify attention.

4. Whether low CTR may be related to the query, search intent, title, snippet, or another factor not represented in the model.

5. Whether low engagement reflects a genuine content issue or a measurement or page type difference.

6. Whether the page is strategically important to the client.

7. Whether the suggested action is appropriate after reading the actual content.

The model provides a priority signal. The final decision remains with the human reviewer.

### No go list

The ranked queue must not be used for the following decisions without human review:

1. Do not automatically edit, refresh, delete, publish, or unpublish content.

2. Do not treat the model probability as proof that a page is declining.

3. Do not claim that a recommended refresh will increase rankings, traffic, CTR, sessions, or engagement.

4. Do not make decisions using a reason code alone. Review the page and the supporting metrics together.

5. Do not automatically change titles, headings, keywords, search snippets, or page structure based only on the model recommendation.

6. Do not treat low CTR or low engagement as proof of poor content quality.

7. Do not compare clients as if their traffic levels, history, audiences, and business goals are identical.

8. Do not use pseudonymous content or client identifiers as predictive signals.

9. Do not bypass editorial, SEO, legal, brand, or client approval requirements.

10. Do not use the system as an autonomous decision maker. It is a prioritization tool for human reviewers.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring and retrain triggers

The current model and action queue should not be treated as permanently valid. Content performance, search demand, client portfolios, and page characteristics can change over time.

### What should be monitored

1. **Precision at 50**

The current Random Forest achieved a precision at 50 of 0.740 on the client holdout validation. Future validation runs should compare against this reference value.

2. **Model score distribution**

Track whether predicted probabilities shift substantially compared with the current scoring run. A large shift may indicate that the new content population differs from the data used to build the model.

3. **Action distribution**

Track the percentage of pages assigned to `monitor`, `refresh`, `refresh_and_review_ctr`, `refresh_and_review_engagement`, and `expand_and_refresh`. A large change in this mix should be reviewed before using the queue operationally.

4. **Confidence distribution**

Track the number and percentage of high, medium, and low confidence recommendations. A sudden reduction in high confidence items may indicate weaker model evidence or changed input data.

5. **Input data quality**

Check missing values, unexpected values, new content types, unusual traffic distributions, and changes in important model features before producing a new queue.

6. **Reason code distribution**

Monitor how frequently each reason code appears. Large changes may indicate changes in the content portfolio or data collection process.

### Retrain or investigate when

1. Precision at 50 falls materially below the current reference value of 0.740.

2. Model probability or important feature distributions change substantially from the reference data.

3. A large change appears in the action or confidence distribution.

4. New content types or data patterns appear that were not represented adequately during training.

5. Data collection definitions or source systems change.

6. Repeated human review shows that highly ranked recommendations are no longer useful.

7. A new labelled dataset becomes available that represents more recent content behaviour.

A trigger does not automatically mean the model must be replaced. It means the model, data pipeline, and validation results should be investigated before the recommendations continue to be used.

```python
monitoring_reference = {
    "precision_at_50": 0.740,
    "queue_rows": len(queue),
    "mean_model_probability": queue["best_model_probability"].mean(),
    "median_model_probability": queue["best_model_probability"].median(),
    "mean_refresh_score": queue["final_refresh_score"].mean(),
    "median_refresh_score": queue["final_refresh_score"].median(),
}

print("=" * 70)
print("CURRENT MONITORING REFERENCE")
print("=" * 70)

for metric, value in monitoring_reference.items():
    if isinstance(value, float):
        print(f"{metric}: {value:.3f}")
    else:
        print(f"{metric}: {value}")

print("\nACTION MIX")
action_monitor = (
    queue["suggested_action"]
    .value_counts()
    .rename("count")
    .to_frame()
)

action_monitor["percentage"] = (
    action_monitor["count"] / len(queue) * 100
).round(1)

display(action_monitor)

print("\nCONFIDENCE MIX")
confidence_monitor = (
    queue["confidence"]
    .value_counts()
    .rename("count")
    .to_frame()
)

confidence_monitor["percentage"] = (
    confidence_monitor["count"] / len(queue) * 100
).round(1)

display(confidence_monitor)

print("\nMODEL PROBABILITY QUANTILES")
print(
    queue["best_model_probability"]
    .quantile([0.25, 0.50, 0.75])
    .round(3)
)

print("\nREFRESH SCORE QUANTILES")
print(
    queue["final_refresh_score"]
    .quantile([0.25, 0.50, 0.75])
    .round(3)
)
```
